validation the score

```python
name = input("Enter your name").strip().title()
score = float(input("Enter your score(0-100): "))
print()
print("\n======YOUR RESULTS=======\n")
print()

if score < 0 or score > 100:
    print("Invalid score. Enter a value between 0 and 100")
elif score >= 70:
    print("Grade A")
elif score >= 60:
    print("Grade B")
elif score >= 50:
    print("Grade C")
elif score >= 45:
    print("Grade D")
else:
    print("Grade F")
```

logical operator (or)
either condition is true

```python
order_total = float(input("Enter your order total: "))
has_membership = input("Membership card? (yes/no): ").strip().lower()
if order_total >= 20000 or has_membership == "yes":
    print("You qualify for free delivery")
else:
    print("Delivery fee applies")
```

logical operator (not)

```python
is_raining = False
if not is_raining:
    print("It is not raining")
else:
    print("It is raining. Take an umbrella")
```

checking whether value is within range

```python
temp = float(input("Enter your temperature: "))
if temp >= 20 and temp <= 30:
    print("Temperature is in range 20-30")
else:
    print("Temperature is outside the  range 20-30")
```

nested if statements

```python
age = int(input("Enter your age"))
has_ticket = input("Do you have a ticket? (yes/no): ").strip().lower()
if age >= 18:
    if has_ticket == "yes":
        print("Entry permitted")
    else:
        print("You need a ticket")
else:
    print("You dont meet the age requirement")
```
