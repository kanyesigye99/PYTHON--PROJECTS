## 1. What is a collection?

A collection stores multiple values inside one variable. For example, instead of creating separate variables for every student's name, we can keep all the names together.

Python has several collection types. The main ones in this lesson are:
- **List:** ordered and changeable; duplicates allowed.
- **Tuple:** ordered and not changeable; duplicates allowed.
- **Set:** stores unique items and does not support positional indexing.
- **Dictionary:** stores information as key-value pairs.

````python
student1 = "Ada"
student2 = "Tunde"
student3 = "Mariam"

students = ["Ada", "Tunde", "Mariam"]
print(students)
````

## 2. Lists

A list is an ordered, changeable collection. Lists use square brackets `[]`. They can contain duplicate values, and they can contain different data types, though related items are often easier to manage.

````python
names = ["Ada", "Tunde", "Mariam"]
numbers = [10, 20, 30, 40]
mixed = ["Janet", 25, True, 3.5]

print(names)
print(numbers)
print(mixed)
````

### 2.1 Access items with indexes

Python indexes start at **0**. Index `0` is the first item, index `1` is the second, and index `-1` is the last. An index that does not exist causes an `IndexError`.

````python
fruits = ["mango", "orange", "banana", "apple"]
print(fruits[0])
print(fruits[1])
print(fruits[-1])
print(fruits[-2])
````

### 2.2 Count items with `len()`

`len()` returns the number of items. A list with four items has indexes `0` through `3`, so its last positive index is `len(list_name) - 1`.

````python
fruits = ["mango", "orange", "banana", "apple"]
print("Number of fruits:", len(fruits))
print("Last valid positive index:", len(fruits) - 1)
````

### 2.3 Slicing

Slicing retrieves a section of a list using `list_name[start:stop]`. The start is included, but the stop is excluded. You can omit the start or stop. A third value specifies the step.

````python
numbers = [0, 1, 2, 3, 4, 5, 6, 7]
print(numbers[1:4])
print(numbers[:3])
print(numbers[4:])
print(numbers[::2])
print(numbers[::-1])
````

### 2.4 Change an item

Lists are mutable, which means you can replace an item after creating the list.

````python
colors = ["red", "blue", "green"]
print("Before:", colors)
colors[1] = "purple"
print("After:", colors)
````

### 2.5 Add items

- `append(item)` adds one item to the end.
- `insert(index, item)` adds an item at a particular position.
- `extend(iterable)` adds each item from another iterable.

**Difference:** `append([3, 4])` adds the whole list as one nested item; `extend([3, 4])` adds `3` and `4` as separate items.

````python
fruits = ["mango", "orange"]
fruits.append("banana")
print("After append:", fruits)
fruits.insert(1, "pawpaw")
print("After insert:", fruits)
fruits.extend(["apple", "grape"])
print("After extend:", fruits)
````

````python
first = [1, 2]
first.append([3, 4])
print("append:", first)

second = [1, 2]
second.extend([3, 4])
print("extend:", second)
````

### 2.6 Remove items

- `remove(value)` removes the first matching value. It raises `ValueError` if the value is missing.
- `pop(index)` removes and returns an item; with no index, it removes the last item.
- `del list_name[index]` deletes an item at a position.
- `clear()` removes all items.

````python
fruits = ["mango", "orange", "banana", "orange"]
fruits.remove("orange")
print("After remove:", fruits)

removed_item = fruits.pop()
print("Popped item:", removed_item)
print("After pop:", fruits)

del fruits[0]
print("After del:", fruits)

fruits.clear()
print("After clear:", fruits)
````

### 2.7 Useful list methods and functions

- `sort()` sorts the existing list in place.
- `sorted(list)` returns a new sorted list without changing the original.
- `reverse()` reverses the existing list.
- `count(value)` counts occurrences.
- `index(value)` gives the first matching index.
- `min()`, `max()`, and `sum()` are useful for numeric lists.

````python
scores = [65, 90, 72, 90, 48]
print("Count of 90:", scores.count(90))
print("Index of 72:", scores.index(72))
print("Lowest:", min(scores))
print("Highest:", max(scores))
print("Total:", sum(scores))
print("Average:", sum(scores) / len(scores))

scores.sort()
print("Sorted in place:", scores)
descending = sorted(scores, reverse=True)
print("New descending list:", descending)
print("Original list:", scores)
````

### 2.8 Check whether an item exists

The `in` and `not in` operators check whether an item is present. They return Boolean values: `True` or `False`.

````python
students = ["Ada", "Tunde", "Mariam"]
print("Ada" in students)
print("Bola" in students)
print("Bola" not in students)
````

### 2.9 Loop through a list

A `for` loop processes each item. Use `enumerate()` if you need both the item's position and its value.

````python
students = ["Ada", "Tunde", "Mariam"]

for student in students:
    print("Welcome,", student)

for number, student in enumerate(students, start=1):
    print(number, student)
````

## 3. Tuples

A tuple is an ordered collection commonly written with parentheses `()`. Tuples are **immutable**: their items cannot be reassigned after creation. They allow duplicates. They are useful for values that belong together and should not be changed accidentally, such as coordinates.

````python
coordinates = (6.5, 3.2)
rgb_color = (255, 120, 0)
days = ("Monday", "Tuesday", "Wednesday")

print(coordinates[0])
print(rgb_color)
print(len(days))
print("Tuesday" in days)
````

### 3.1 A one-item tuple

The comma makes a one-item tuple. `(5)` is just the integer `5`; `(5,)` is a tuple.

````python
not_a_tuple = (5)
one_item_tuple = (5,)

print(type(not_a_tuple))
print(type(one_item_tuple))
````

### 3.2 Tuple unpacking

Tuple unpacking assigns tuple items to separate variables. Normally, the number of variables must match the number of items.

````python
student = ("Mariam", 19, "Chemistry")
name, age, course = student

print(name)
print(age)
print(course)
````

## 4. Sets

A set stores **unique** items. Sets are useful for removing duplicates and comparing groups. They do not support positional indexing, and you should not rely on their display order. Items must be hashable; strings and integers are common choices.

````python
departments = {"Science", "Arts", "Science", "Technology"}
print(departments)
print("Unique departments:", len(departments))
````

### 4.1 Add and remove set items

- `add(item)` adds one item.
- `update(iterable)` adds several items.
- `discard(item)` removes an item if present and does not fail if absent.
- `remove(item)` raises `KeyError` if the item is absent.

````python
languages = {"Python", "JavaScript"}
languages.add("SQL")
print(languages)
languages.update(["HTML", "CSS"])
print(languages)
languages.discard("Java")
languages.remove("SQL")
print(languages)
````

### 4.2 Set operations

Given two sets:
- **Union (`|`)**: items in either set.
- **Intersection (`&`)**: items in both sets.
- **Difference (`-`)**: items in the first set but not the second.
- **Symmetric difference (`^`)**: items in either set but not both.

````python
python_students = {"Ada", "Tunde", "Mariam", "Bola"}
javascript_students = {"Mariam", "Bola", "David"}

print("Either course:", python_students | javascript_students)
print("Both courses:", python_students & javascript_students)
print("Python only:", python_students - javascript_students)
print("Only one course:", python_students ^ javascript_students)
````

## 5. Dictionaries

A dictionary stores information as **key-value pairs**. It uses curly braces `{}` with a colon between a key and its value. For example, `"name"` can be a key and `"Ada"` its value.

Keys must be unique and hashable. Values can have different data types. Dictionaries are changeable and preserve insertion order in modern Python.

````python
student = {
    "name": "Ada",
    "age": 18,
    "course": "Python",
    "score": 87
}

print(student)
print(student["name"])
print(student["score"])
````

### 5.1 Access values safely

`student["name"]` retrieves a value but raises `KeyError` if the key is missing. `get()` can return a default value when the key is absent.

````python
student = {"name": "Ada", "score": 87}
print(student["name"])
print(student.get("course"))
print(student.get("course", "Not registered"))
````

### 5.2 Add and update dictionary values

Assigning to an existing key updates its value. Assigning to a new key creates it. `update()` can add or update multiple pairs at once.

````python
student = {"name": "Ada", "score": 75}
student["score"] = 92
student["course"] = "Python"
student.update({"city": "Akure", "active": True})
print(student)
````

### 5.3 Remove dictionary items

`pop(key)` removes the key and returns its value. `del dictionary[key]` deletes a pair. `clear()` removes all pairs.

````python
student = {"name": "Ada", "score": 92, "course": "Python"}
score = student.pop("score")
print("Removed score:", score)
print(student)
del student["course"]
print(student)
````

### 5.4 Loop through a dictionary

`keys()` gives the keys, `values()` gives the values, and `items()` gives key-value pairs. Looping over the dictionary directly also returns its keys.

````python
student = {"name": "Ada", "age": 18, "course": "Python"}

for key in student:
    print(key)

for value in student.values():
    print(value)

for key, value in student.items():
    print(f"{key}: {value}")
````

## 6. Choosing the right collection

| Collection | Syntax | Ordered? | Changeable? | Duplicates? | Best use |
|---|---|---|---|---|---|
| List | `[]` | Yes | Yes | Yes | An editable sequence |
| Tuple | `()` | Yes | No | Yes | A fixed group of values |
| Set | `{1, 2}` or `set()` | No positional order | Yes | No | Unique items and set operations |
| Dictionary | `{"key": "value"}` | Insertion order | Yes | Keys are unique | Labeled data and records |

Empty collections:
- Empty list: `[]`
- Empty tuple: `()`
- Empty set: `set()`
- Empty dictionary: `{}` — note that `{}` is not an empty set.

````python
empty_list = []
empty_tuple = ()
empty_set = set()
empty_dictionary = {}

print(type(empty_list))
print(type(empty_tuple))
print(type(empty_set))
print(type(empty_dictionary))
````

## 7. Practical Project 1 — Student Scores Manager

This project stores scores in a list and calculates the number of scores, total, average, highest, and lowest. This example assumes the list contains at least one score.

````python
scores = [78, 85, 92, 66, 88]
total = sum(scores)
average = total / len(scores)

print("Scores:", scores)
print("Number of students:", len(scores))
print("Total score:", total)
print(f"Average score: {average:.2f}")
print("Highest score:", max(scores))
print("Lowest score:", min(scores))
````

### Project 1 extension — Grade each score

This combines collections with `for` loops and `if` statements from the previous lesson.

````python
scores = [78, 35, 92, 49, 66, 88]

for score in scores:
    if score >= 70:
        grade = "A"
    elif score >= 60:
        grade = "B"
    elif score >= 50:
        grade = "C"
    elif score >= 45:
        grade = "D"
    else:
        grade = "F"
    print(f"Score: {score} | Grade: {grade}")
````

## 8. Practical Project 2 — Student Profile

A dictionary is useful when one record has labeled fields. Here, the student's scores are stored in a list inside the dictionary, and we calculate an average.

````python
student = {
    "name": "Mariam",
    "age": 20,
    "course": "Python",
    "scores": [80, 91, 76]
}

average = sum(student["scores"]) / len(student["scores"])
student["average"] = round(average, 2)

for key, value in student.items():
    print(f"{key}: {value}")
````

## 9. Practical Project 3 — Remove Duplicate Names

A set removes duplicates, but the display order is not guaranteed. If you want to remove duplicates while preserving the original order, `dict.fromkeys()` is a useful option.

````python
names = ["Ada", "Mariam", "Tunde", "Ada", "Bola", "Mariam"]

unique_names = set(names)
unique_names_in_order = list(dict.fromkeys(names))

print("Unique names:", unique_names)
print("Unique names in original order:", unique_names_in_order)
````

## 10. Common mistakes

1. **Wrong index:** Python list indexes begin at `0`.
2. **Index out of range:** check the list length before accessing a position.
3. **Confusing `append()` and `extend()`:** one adds a single item; the other adds each item from an iterable.
4. **Expecting `sort()` to return a new list:** it changes the list and returns `None`.
5. **Trying to change a tuple:** tuples are immutable.
6. **Using `{}` for an empty set:** it creates an empty dictionary; use `set()`.
7. **Accessing a missing dictionary key:** use `get()` when the key may not exist.
8. **Expecting set items to have positional indexes:** sets do not support indexing.

````python
numbers = [4, 2, 8]
result = numbers.sort()
print("Sorted list:", numbers)
print("Return value of sort():", result)
````
