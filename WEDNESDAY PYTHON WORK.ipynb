#outer, inner and the for loop

```python
i = 1
while i <= 5: #outer loop
    for j in range(i): #inner loop
        print(i, end="") #Belongs to the for loop

    print()
    i = i + 1
```

#How to check even or odd numbers

```python
i = 1
while i <= 10:
    if i % 2 ==0:
        print(i, "is even")
    else:
        print(i, "is odd")
    i = i + 1
```

```python
i = 1
while i <= 4:
    j = 1
    while j <= i:
        print("*", end="")
        j = j + 1

    print()
    i = i + 1
```
