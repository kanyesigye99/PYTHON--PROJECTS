1. Create a function. The function is called self defined which has a parameterist of a and b that divides a by b. If b is zero handle the zero division error and return none instead of letting the program crash.
2. Create a list, ask a user to enter an index and print an element at that index. If the index is outside the valid range, catch the index error and display an informative message.
3. Given a dictionary of country capitals, ask the user to enter a country name and print the corresponding capital. If the country is not in the dictionary, catch the key error and display a helpful message.
4. Write a function and values with parameterist of a and b that returns the sum of two values. If the argument is of incompatible types, catch the type error and return a descriptive error message instead of program crushing.

```python
#No: 1
def self_defined(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        return "Cant divide by 0"
self_defined(20,0)
```

```python
#No: 2
items = ["bandminton","baseball","golf","cricket","dancesport", "swimming","pickleball"]
index = int(input(f"Write an index from 0 to 6: "))
try:
    print(items[index])
except IndexError:
    print(f"Invalid index. Enter a number from 0 to 6.")
```

```python
#No: 3
capitals = {
    "Congo":"Kishansha", "Algeria": "Algiers", "Kenya": "Nairobi","Tanzania": "Dar-es-salaam",
     "Rwanda": "Kigali", "Egypt": "Cairo", "South Sudan":"Juba"}
country = input("Enter the country name: ")

try:
    print(f"The capital city of {country} is {capitals[country]}.")
except KeyError:
    print(f"Sorry, '{country}' is not in the dictionary.")
```

```python
#No: 4
def add_values(a, b):
    try:
        return a + b
    except TypeError:
        return "Error: a and b should be of compatible types to add."
print(add_values(100, 200))      
```
